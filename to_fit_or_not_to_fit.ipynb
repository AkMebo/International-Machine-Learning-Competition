```python
import numpy as np
```

```python
x = np.array([0, 1, 2, 3])
y = np.array([1, 3.2, 4.8, 7])
lam = 1
```

```python
# Model 1
# Note: The problem gives the equation in descending order, we must put coefficients in order [x^3, x^2, x, intercept]
coeffs_m1 = [0.2, -0.9, 2.9, 1]
preds_m1 = np.polyval(coeffs_m1, x)
mse_m1 = np.sum((y - preds_m1)**2)
penalty_m1 = lam * np.sum(np.array(coeffs_m1[:3])**2) # Exclude intercept from penalty
J1 = mse_m1 + penalty_m1
print(f"Model 1 Score: {J1:.2f}")
```

```python
# Model 2
# Equation: 2x + 1. Coeffs are [x, intercept]
coeffs_m2 = [2, 1]
preds_m2 = np.polyval(coeffs_m2, x)
mse_m2 = np.sum((y - preds_m2)**2)
penalty_m2 = lam * np.sum(np.array(coeffs_m2[:1])**2) # Exclude intercept
J2 = mse_m2 + penalty_m2
print(f"Model 2 Score: {J2:.2f}")
```
